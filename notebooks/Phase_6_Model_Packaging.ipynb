```python
# ============================================================
# PHASE 6.2 — UNIFIED PREDICTION FUNCTION
# ============================================================

import pandas as pd
import numpy as np

def predict_project_row(project_row):
    """
    Takes one raw project feature row and returns
    all downstream model outputs.
    """

    # --------------------------------------------------------
    # 1. Convert input into one-row DataFrame
    # --------------------------------------------------------
    if isinstance(project_row, pd.Series):
        row = project_row.to_frame().T.copy()
    elif isinstance(project_row, dict):
        row = pd.DataFrame([project_row])
    elif isinstance(project_row, pd.DataFrame):
        row = project_row.copy()
    else:
        raise TypeError("project_row must be a dict, Series, or DataFrame")

    # --------------------------------------------------------
    # 2. Make sure all 47 XGBoost features exist
    # --------------------------------------------------------
    missing_features = [
        col for col in feature_columns
        if col not in row.columns
    ]

    if missing_features:
        raise ValueError(
            f"Missing required features: {missing_features}"
        )

    xgb_row = row[feature_columns].copy()

    # --------------------------------------------------------
    # 3. Cost risk
    # --------------------------------------------------------
    cost_probability = float(
        cost_final_model.predict_proba(xgb_row)[0, 1]
    )

    # --------------------------------------------------------
    # 4. Schedule risk
    # --------------------------------------------------------
    schedule_probability = float(
        schedule_final_model.predict_proba(xgb_row)[0, 1]
    )

    # --------------------------------------------------------
    # 5. Cox risk
    # --------------------------------------------------------
    cox_row = row[cox_feature_columns].copy()

    # Cox model cannot accept missing values.
    # Use the model's training means for missing inputs.
    for col in cox_feature_columns:
        if cox_row[col].isna().any():
            if hasattr(cox_model, "_norm_mean"):
                fill_value = float(cox_model._norm_mean[col])
            else:
                fill_value = 0.0

            cox_row[col] = cox_row[col].fillna(fill_value)

    cox_risk = float(
        cox_model.predict_partial_hazard(cox_row).iloc[0]
    )

    # --------------------------------------------------------
    # 6. Normalize Cox risk approximately to 0-1
    # --------------------------------------------------------
    # Partial hazard itself is not a probability.
    # For the unified score, convert it to a bounded value.
    cox_risk_probability = float(
        cox_risk / (1.0 + cox_risk)
    )

    # --------------------------------------------------------
    # 7. Composite risk score
    # --------------------------------------------------------
    composite_score = (
        0.40 * cost_probability +
        0.40 * schedule_probability +
        0.20 * cox_risk_probability
    ) * 100

    composite_score = float(
        np.clip(composite_score, 0, 100)
    )

    # --------------------------------------------------------
    # 8. Risk tier
    # --------------------------------------------------------
    if composite_score < 25:
        risk_tier = "LOW"
    elif composite_score < 50:
        risk_tier = "WATCH"
    elif composite_score < 75:
        risk_tier = "ELEVATED"
    else:
        risk_tier = "CRITICAL"

    # --------------------------------------------------------
    # 9. Return backend-friendly dictionary
    # --------------------------------------------------------
    return {
        "cost_risk_probability": round(cost_probability, 6),
        "schedule_risk_probability": round(schedule_probability, 6),
        "cox_risk": round(cox_risk, 6),
        "cox_risk_probability": round(cox_risk_probability, 6),
        "composite_risk_score": round(composite_score, 4),
        "risk_tier": risk_tier,
        "model_version": package_metadata["package_version"]
        if "package_metadata" in globals()
        else "1.0.0"
    }

print("✅ predict_project_row() created")
```

```python
def manual_test_project(project_row, top_n=8):
    """
    Manual project tester with:
    - Project information
    - Cost risk
    - Schedule risk
    - Cox survival risk
    - Composite risk score
    - Risk tier
    - Model interpretation
    - SHAP explanation for Cost XGBoost
    - SHAP explanation for Schedule XGBoost

    project_row can be:
        - pandas Series
        - dict
        - one-row DataFrame
    """

    import numpy as np
    import pandas as pd
    import shap

    # ============================================================
    # 1. NORMALIZE INPUT
    # ============================================================

    if isinstance(project_row, pd.DataFrame):
        if len(project_row) != 1:
            raise ValueError("project_row DataFrame must contain exactly one row.")
        row = project_row.iloc[0]

    elif isinstance(project_row, pd.Series):
        row = project_row

    elif isinstance(project_row, dict):
        row = pd.Series(project_row)

    else:
        raise TypeError(
            "project_row must be a pandas Series, dict, or one-row DataFrame."
        )

    # ============================================================
    # 2. BASIC PROJECT INFORMATION
    # ============================================================

    project_id = row.get("project_id", "N/A")
    sector = row.get("sector", "N/A")
    state = row.get("state", "N/A")
    report_month = row.get("report_month", "N/A")
    current_cost = row.get("current_cost", np.nan)
    expenditure = row.get("cumulative_expenditure_crore", np.nan)

    print("\n" + "=" * 65)
    print("AI INFRASTRUCTURE RISK MODEL — MANUAL TESTER")
    print("=" * 65)

    print("\n---")
    print("PROJECT INFORMATION")
    print("\n")

    print(f"Project ID       : {project_id}")
    print(f"Sector           : {sector}")
    print(f"State            : {state}")
    print(f"Report Month     : {report_month}")

    if pd.notna(current_cost):
        print(f"Current Cost     : ₹{float(current_cost):,.2f} crore")
    else:
        print("Current Cost     : N/A")

    if pd.notna(expenditure):
        print(f"Expenditure      : ₹{float(expenditure):,.2f} crore")
    else:
        print("Expenditure      : N/A")

    # ============================================================
    # 3. RUN MAIN PREDICTION
    # ============================================================

    print("\n" + "=" * 65)
    print("RUNNING MODELS...")
    print("=" * 65)

    prediction = predict_project_row(row)

    cost_prob = float(prediction["cost_risk_probability"])
    schedule_prob = float(prediction["schedule_risk_probability"])
    cox_risk = float(prediction["cox_risk"])
    cox_prob = float(prediction["cox_risk_probability"])
    composite = float(prediction["composite_risk_score"])
    risk_tier = prediction["risk_tier"]
    model_version = prediction["model_version"]

    # ============================================================
    # 4. MODEL PREDICTION
    # ============================================================

    print("\n" + "=" * 65)
    print("MODEL PREDICTION")
    print("=" * 65)

    print(f"\n💰 Cost Overrun Risk     : {cost_prob:.2%}")
    print(f"📅 Schedule Risk         : {schedule_prob:.2%}")
    print(f"📈 Cox Survival Risk     : {cox_prob:.2%}")

    print("\n" + "-" * 32)
    print(f"\n🔥 COMPOSITE RISK SCORE  : {composite:.2f} / 100")
    print(f"🚦 RISK TIER             : {risk_tier}")

    # ============================================================
    # 5. MODEL INTERPRETATION
    # ============================================================

    print("\n" + "-" * 32)
    print("\nMODEL INTERPRETATION")
    print("=" * 32)

    if cost_prob >= 0.50:
        print("⚠️ HIGH probability of future cost overrun.")
    elif cost_prob >= 0.25:
        print("🟡 MODERATE probability of future cost overrun.")
    else:
        print("🟢 LOW probability of future cost overrun.")

    if schedule_prob >= 0.50:
        print("⚠️ HIGH probability of future schedule risk.")
    elif schedule_prob >= 0.25:
        print("🟡 MODERATE probability of future schedule risk.")
    else:
        print("🟢 LOW probability of future schedule risk.")

    # ============================================================
    # 6. SHAP HELPER
    # ============================================================

    def get_shap_explanation(pipeline, feature_names, row, model_name):
        """
        Calculate SHAP values for one row from a sklearn Pipeline
        containing a ColumnTransformer + XGBoost model.
        """

        # --------------------------------------------------------
        # Find preprocessing and model components
        # --------------------------------------------------------

        if hasattr(pipeline, "named_steps"):
            steps = pipeline.named_steps

            preprocessor = None
            model = None

            for name, step in steps.items():

                # ColumnTransformer
                if hasattr(step, "transformers_"):
                    preprocessor = step

                # XGBoost classifier
                if hasattr(step, "get_booster"):
                    model = step

            if preprocessor is None or model is None:
                raise ValueError(
                    f"Could not identify preprocessor/model in {model_name} pipeline."
                )

        else:
            raise ValueError(
                f"{model_name} model is not a sklearn Pipeline."
            )

        # --------------------------------------------------------
        # Build input DataFrame with exact feature order
        # --------------------------------------------------------

        X_row = pd.DataFrame(
            [[row.get(col, np.nan) for col in feature_names]],
            columns=feature_names
        )

        # --------------------------------------------------------
        # Transform using the SAME preprocessing used in training
        # --------------------------------------------------------

        X_transformed = preprocessor.transform(X_row)

        # --------------------------------------------------------
        # Get transformed feature names
        # --------------------------------------------------------

        transformed_names = preprocessor.get_feature_names_out()

        # --------------------------------------------------------
        # Convert sparse matrix if necessary
        # --------------------------------------------------------

        if hasattr(X_transformed, "toarray"):
            X_transformed = X_transformed.toarray()

        X_transformed = np.asarray(X_transformed)

        # --------------------------------------------------------
        # SHAP
        # --------------------------------------------------------

        explainer = shap.TreeExplainer(model)

        shap_values = explainer.shap_values(X_transformed)

        # XGBoost binary classification normally returns:
        # shape = (1, n_features)

        if isinstance(shap_values, list):
            shap_values = shap_values[-1]

        shap_values = np.asarray(shap_values)

        if shap_values.ndim == 1:
            values = shap_values
        else:
            values = shap_values[0]

        # --------------------------------------------------------
        # Clean feature names
        # --------------------------------------------------------

        clean_names = []

        for name in transformed_names:

            if name.startswith("num__"):
                clean_name = name.replace("num__", "", 1)

            elif name.startswith("cat__"):
                clean_name = name.replace("cat__", "", 1)

            else:
                clean_name = name

            clean_names.append(clean_name)

        # --------------------------------------------------------
        # Create SHAP table
        # --------------------------------------------------------

        shap_df = pd.DataFrame({
            "feature": clean_names,
            "shap_value": values,
            "abs_shap": np.abs(values)
        })

        shap_df = shap_df.sort_values(
            "abs_shap",
            ascending=False
        ).head(top_n)

        # Direction
        shap_df["direction"] = np.where(
            shap_df["shap_value"] > 0,
            "increases_risk",
            "decreases_risk"
        )

        return shap_df

    # ============================================================
    # 7. COST SHAP
    # ============================================================

    print("\n" + "=" * 65)
    print("SHAP EXPLANATION — COST RISK")
    print("=" * 65)

    try:

        cost_shap = get_shap_explanation(
            cost_final_model,
            feature_columns,
            row,
            "Cost XGBoost"
        )

        for _, r in cost_shap.iterrows():

            symbol = "🔴" if r["shap_value"] > 0 else "🟢"

            print(
                f"{symbol} "
                f"{r['feature']:<40} "
                f"{r['shap_value']:+.4f}  →  "
                f"{r['direction']}"
            )

    except Exception as e:

        print(f"⚠️ Cost SHAP explanation unavailable: {e}")

    # ============================================================
    # 8. SCHEDULE SHAP
    # ============================================================

    print("\n" + "=" * 65)
    print("SHAP EXPLANATION — SCHEDULE RISK")
    print("=" * 65)

    try:

        schedule_shap = get_shap_explanation(
            schedule_final_model,
            feature_columns,
            row,
            "Schedule XGBoost"
        )

        for _, r in schedule_shap.iterrows():

            symbol = "🔴" if r["shap_value"] > 0 else "🟢"

            print(
                f"{symbol} "
                f"{r['feature']:<40} "
                f"{r['shap_value']:+.4f}  →  "
                f"{r['direction']}"
            )

    except Exception as e:

        print(f"⚠️ Schedule SHAP explanation unavailable: {e}")

    # ============================================================
    # 9. RAW OUTPUT
    # ============================================================

    print("\n" + "=" * 65)
    print("RAW PREDICTION OUTPUT")
    print("=" * 65)

    print(prediction)

    print("\n" + "=" * 65)
    print("✅ Prediction + SHAP explanation completed successfully.")
    print("=" * 65)

    # ============================================================
    # 10. RETURN EVERYTHING
    # ============================================================

    return {
        "prediction": prediction,
        "cost_shap": cost_shap if "cost_shap" in locals() else None,
        "schedule_shap": schedule_shap if "schedule_shap" in locals() else None
    }
```
